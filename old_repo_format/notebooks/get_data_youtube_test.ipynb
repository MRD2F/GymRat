````python
ytt_api = YouTubeTranscriptApi()

def get_transcript(video_id):
    #video_id="pxiP-HJLCx0"
    transcript = ytt_api.fetch(video_id, languages=['en'])
    text = " ".join([x.text for x in transcript])
    return text
````

## ✅ 4. **Create a Dataset in Instruction Format (for LLM training)**

Format each example like this:

```json
{
  "instruction": "Explain how to properly perform the deadlift.",
  "input": "",
  "output": "Start by standing with feet hip-width apart, grip the bar just outside the knees..."
}
```

Other sample prompts:

* "What are the common mistakes when doing squats?"
* "Write a 4-week hypertrophy program."
* "What supplements help with recovery?"

* **Starting Strength (Mark Rippetoe)** – (if you own a copy)
