```python
def load_dataset(json_file_name):
    with open(json_file_name, "r", encoding="utf-8") as f:
        dataset = json.load(f)
    return dataset

def format_dataset(dataset):
    clean_dataset = [
        {
            "content": item['prompt'][1]['content'],
            "instruction": item["instruction"] ,  # user question
            "output": item["output"]        # model's response
        }
        for item in dataset
        if "prompt" in item and len(item["prompt"]) > 1  # sanity check
    ]
    return clean_dataset
```

```python
def is_valid(text):
    if pd.isna(text):  
        return False
    if not isinstance(text, str):  # Check if not a string
        return False

    text = text.strip().lower()
    # Check for known invalid values or if 'null' is present anywhere
    invalid_keywords =["null", "none", "nan", "n/a"] 
    if any(bad in text for bad in invalid_keywords):
        return False

    #minimum content length
    if len(text) < 5:  
        return False
    return True

def filter_output(df):
    # Apply filter to both columns
    #print(df.head())
    filtered_df = df[df['instruction'].apply(is_valid) & df['output'].apply(is_valid)].copy()
    filtered_df = filtered_df.drop_duplicates()

    cleaned_data = filtered_df.to_dict(orient="records")
    return cleaned_data

def clean_json_file(dataset):
    df = pd.DataFrame(dataset)
    print('*********',df.head())
    n_initial_samples = df.shape[0]

    #returns a json format out of the input df after being preprocessed
    cleaned_data = filter_output(df)
    n_records = len(cleaned_data)
    #json_output_name = f"synthetic_QA_{datetime.now().strftime('%Y%m%d_%H%M%S')}.json"
        
    # with open(f"{json_output_name}", "w", encoding="utf-8") as f:
    #     json.dump(cleaned_data, f, ensure_ascii=False, indent=2)

    return cleaned_data
```

```python
def get_and_save_cleaned_dataset(json_file_name):
    dataset = load_dataset(json_file_name)
    dataset_formatted = format_dataset(dataset)
    clean_dataset = clean_json_file(dataset_formatted)

    json_output_name = f"cleaned_{json_file_name}"
        
    with open(f"{json_output_name}", "w", encoding="utf-8") as f:
        json.dump(clean_dataset, f, ensure_ascii=False, indent=2)


    return clean_dataset
```
