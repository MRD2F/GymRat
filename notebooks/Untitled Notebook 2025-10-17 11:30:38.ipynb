```python
import json
```

```python
json_output_name= f"test_16-10-25"
with open(f"{json_output_name}.json", "r") as f:
    results = json.load(f)
```

```python
def get_key(d, n, key):
    print(d[n][key])
```

```python
for n in range(60,90):
    if n%3==0:
        print('*'*15,int(n/3),'*'*15)
        print(results[n]['prompt'][1]['content'])
        
    print('-'*15, n, '-'*15)
    get_key(results, n, "instruction")
    get_key(results, n, "output")
```
